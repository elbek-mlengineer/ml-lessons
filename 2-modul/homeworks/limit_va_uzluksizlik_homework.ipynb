```python
def taxi(x):
  return 3000*x + 5000


chapdan = [3.9, 3.99, 3.999]
ongdan = [4.001, 4.01, 4.1]
for x in chapdan:
  print(x, '->', taxi(x))
for x in ongdan:
  print(x, '->', taxi(x))
# bu holatda limit uzluksiz limit bo'ladi, chunki x=4 bo'lganda ham natija beradi
# bunda ikkala tomondan ham 16 ga yaqinlashayotganligi sababli, ikki tomonlama yaqinlashish deyiladi
```

```python
import numpy as np
import matplotlib.pyplot as plt

def cal(x):
  try:
    return (x**2-9)/(x-3)
  except:
    return "x=3 bo'lganligi sababli, sonni 0 ga bo'lib bo'lmaydi."

for x in [2.9, 2.99, 2.999, 3, 3.001, 3.01, 3.1]:
  print(x, '->', cal(x))

# bunda limit 6 ga yaqinlashyapti, bunda limit uzluksiz emas,
# chunki 3 qiymatida nolga bo'lishga to'g'ri kelib qoladi.
# sonni nolga bo'lib bo'lmaydi va o'sha joyda uziladi.
xs = np.linspace(0, 4, 400)
xs = xs[xs != 3]
plt.scatter(xs, cal(xs), s=2)
plt.scatter(3, 6, s=500, facecolors='none', edgecolors='r')
plt.grid(True)
plt.xlim(0, 4)
plt.ylim(0, 10)
plt.show()
```

```python
# dokon_chegirma.py
import numpy as np
import matplotlib.pyplot as plt
def P(x):
  if x < 500000:
    return x
  else :
    return x * 0.9
chapdan = [499000, 499900, 499990, 499999]
ongdan = [500000, 500001, 500010, 500100, 501000]
for x in chapdan:
  print(x, '->', P(x))
for x in ongdan:
  print(x, '->', P(x))

xs = np.linspace(300000, 700000, 1000)
ys = [P(x) for x in xs]

plt.scatter(xs, ys, s=2)
plt.scatter(500000, P(500000), s=500, facecolors='none', edgecolors='r')
plt.scatter(500000, 500000, s=500, facecolors='none', edgecolors='r')

plt.grid(True)
plt.xlim(300000, 700000)
plt.ylim(250000, 700000)
plt.show()

# bu kodda sakrash mavjud, chunki 500 minggacha chegirma yo'q
# keyin chegirma borligi uchun, birdaniga tushib ketdi.
# bu uzliksiz limit emasligini ko'rsatadi.
```
