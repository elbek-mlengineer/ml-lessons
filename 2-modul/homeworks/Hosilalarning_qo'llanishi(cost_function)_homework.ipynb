```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3])
y = np.array([3, 6, 9])


def cost(w):
    pred = w * x
    return np.mean((pred - y)**2)


def gradient(w):
    pred = w * x
    return np.mean(2 * x * (pred - y))


w = 0
lr = 0.05
epsilon = 1e-6

w_history = [w]
cost_history = [cost(w)]

for step in range(1000):

    g = gradient(w)

    w_new = w - lr * g

    w_history.append(w_new)
    cost_history.append(cost(w_new))

    print(
        f"{step+1}: "
        f"w = {w_new:.6f}, "
        f"cost = {cost(w_new):.6f}"
    )

    # To'xtash sharti
    if abs(w_new - w) < epsilon:
        w = w_new
        print("Gradient descent to'xtadi.")
        break

    w = w_new


# Grafik
w_range = np.linspace(-1, 7, 200)
cost_range = [cost(val) for val in w_range]

plt.figure(figsize=(7, 5))

plt.plot(
    w_range,
    cost_range,
    color="green",
    label="Cost Function"
)

plt.scatter(
    w_history,
    cost_history,
    color="red",
    zorder=5,
    label="Gradient Descent"
)

plt.xlabel("w")
plt.ylabel("Cost")

plt.grid(True)
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3])
y = np.array([3, 6, 9])


def cost(w, b):
    pred = w * x + b
    return np.mean((pred - y)**2)


def gradients(w, b):
    pred = w * x + b

    dw = np.mean(2 * x * (pred - y))
    db = np.mean(2 * (pred - y))

    return dw, db


w = 0
b = 0

lr = 0.05
epsilon = 1e-6

w_history = [w]
b_history = [b]
cost_history = [cost(w, b)]


# Gradient Descent
for step in range(1000):

    # Gradientlarni hisoblash
    dw, db = gradients(w, b)

    # Yangi qiymatlar
    w_new = w - lr * dw
    b_new = b - lr * db

    # Saqlash qoyish qiymatlani
    w_history.append(w_new)
    b_history.append(b_new)
    cost_history.append(cost(w_new, b_new))

    print(
        f"{step+1}: "
        f"w = {w_new:.6f}, "
        f"b = {b_new:.6f}, "
        f"cost = {cost(w_new, b_new):.6f}"
    )

    # To'xtash sharti
    if max(
        abs(w_new - w),
        abs(b_new - b)
    ) < epsilon:

        w = w_new
        b = b_new

        print("\nGradient descent to'xtadi!")
        break

    w = w_new
    b = b_new


print("\nYakuniy natija:")
print(f"w = {w:.6f}")
print(f"b = {b:.6f}")
print(f"cost = {cost(w, b):.6f}")


plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)

plt.plot(
    w_history,
    label="w"
)

plt.plot(
    b_history,
    label="b"
)

plt.xlabel("Iteration")
plt.ylabel("Qiymat")
plt.title("w va b ning yaqinlashishi")

plt.grid(True)
plt.legend()



plt.subplot(1, 2, 2)

plt.plot(
    cost_history,
    color="green"
)

plt.xlabel("Iteration")
plt.ylabel("Cost")
plt.title("Cost kamayishi")

plt.grid(True)

plt.tight_layout()
plt.show()
```
