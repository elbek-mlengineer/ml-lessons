```python
kv = np.array([[0, 1, 1, 0, 0],
               [0, 0, 1, 1, 0]])
Tlar = {'Scalling': [[2, 0], [0, 0.5]],
        'Rotation': [[0, -1], [1, 0]],
        'Reflection': [[1, 0], [0, -1]],
        'Shear': [[1, 1], [0, 1]]}

fig, axes = plt.subplots(1, 4, figsize=(14, 3.5))
for ax, (nom, T) in zip(axes, Tlar.items()):
  yangi = np.array(T) @ kv
  ax.plot(kv[0], kv[1], 'blue')
  ax.plot(yangi[0], yangi[1], 'red')
  ax.set_title(nom);
  ax.grid(True);
  ax.axis('equal')
```

```python
import numpy as np

R = np.array([
    [0, -1],
    [1,  0]
])

S = np.array([
    [3, 0],
    [0, 1]
])

M = S @ R

print("Yagona matritsa:")
print(M)

x = np.array([1, 0])

print("Natija:")
print(M @ x)
```

```python
import numpy as np

R = np.array([
    [0.707, 0.707],
    [0.707, 0.707]
])

S = np.array([
    [2, 0],
    [0,  1]
])


v = np.array([1, 0])

print(R@S)
print(S@R)
print(R@S@v)
print(S@R@v)
print(np.allclose(R @ S, S @ R))
```

```python
img = np.zeros((8, 8))
img[1:7, 1] = 1
img[1:7, 6] = 1
for i in range(1, 7):
  img[i, i] = 1

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Original Image
axes[0].imshow(img, cmap='gray')
axes[0].set_title('Original Image')
axes[0].axis('off') # Turn off axis labels

# Horizontally flipped image
img_flipped_lr = np.fliplr(img)
axes[1].imshow(img_flipped_lr, cmap='gray')
axes[1].set_title('Horizontally Flipped')
axes[1].axis('off')

# Vertically flipped image
img_flipped_ud = np.flipud(img)
axes[2].imshow(img_flipped_ud, cmap='gray')
axes[2].set_title('Vertically Flipped')
axes[2].axis('off')

plt.tight_layout() # Adjust layout to prevent overlapping titles/labels
plt.show()
```
