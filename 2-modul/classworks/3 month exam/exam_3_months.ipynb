```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

x = np.array([40, 60, 80])
y = np.array([3.2, 3.8, 5.4])

# 1. Uchala kvartira uchun bashorat y ni hisoblash
y_pred = 0.05 * x + 1
print("1. Bashorat qilingan ijara narxlari (y_pred):", y_pred)

# 2. MAE ni topish
mae = mean_absolute_error(y, y_pred)
print(f"2. MAE: {mae:.3f} mln so'm")

# 3. MSE va RMSE ni topish
mse = mean_squared_error(y, y_pred)
rmse = np.sqrt(mse)
print(f"MSE: {mse:.3f}, RMSE: {rmse:.3f} mln so'm")

# 4. R² ni topish (y_bar - o'rtacha qiymat orqali)
y_bar = np.mean(y)
ss_res = np.sum((y - y_pred) ** 2)
ss_tot = np.sum((y - y_bar) ** 2)
r2 = 1 - (ss_res / ss_tot)
print(f"4. R² (qo'lda hisoblangan): {r2:.3f} (y_bar: {y_bar:.2f})")

# 5. w = 0.05 koeffitsiyentining talqini
# Javob:
# w = 0.05 koeffitsiyentining oddiy tilda talqini:
# Kvartira maydoni har 1 metr kvadratga ko'payganda,
# oylik ijara narxi o'rtacha 0.05 mln so'mga (ya'ni 50 000 so'mga) oshadi.

# 6. Modelni o'qitib, test'da R² ni chiqaradigan scikit-learn kodini yozing (3–4 qator)
X_train = x.reshape(-1, 1)
model = LinearRegression().fit(X_train, y)
r2_sklearn = model.score(X_train, y)
print(f"6. Scikit-learn orqali topilgan R²: {r2_sklearn:.3f}")
```

```python
import numpy as np

# Sigmoid funksiyasini aniqlaymiz
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# 1. Ikkala mijoz uchun z va p ni hisoblash
# Mijoz A: x1 = 3, x2 = 2
z_A = 0.8 * 3 - 0.5 * 2 - 1
p_A = sigmoid(z_A)
# Mijoz B: x1 = 0, x2 = 6
z_B = 0.8 * 0 - 0.5 * 6 - 1
p_B = sigmoid(z_B)

print("1. MIJOZLAR UCHUN Z VA P QIYMATLARI:")
print(f"   Mijoz A: z = {z_A:.2f}, p = {p_A:.4f}")
print(f"   Mijoz B: z = {z_B:.2f}, p = {p_B:.4f}\n")

# 2. Chegara 0.5 bo'lganda bashorat qilish
class_A = 1 if p_A >= 0.5 else 0
class_B = 1 if p_B >= 0.5 else 0
print("2. BASHORAT SINF (Chegara = 0.5):")
print(f"   Mijoz A: Sinf {class_A} ({'qaytarmaydi' if class_A == 1 else 'qaytaradi'})")
print(f"   Mijoz B: Sinf {class_B} ({'qaytarmaydi' if class_B == 1 else 'qaytaradi'})\n")

# 3. TP, FP, FN va Recall hisoblash
p_vals = np.array([0.15, 0.35, 0.45, 0.62, 0.81])
y_true = np.array([0, 1, 0, 1, 1])

def evaluate_metrics(p, y, threshold):
    y_pred = (p >= threshold).astype(int)
    tp = np.sum((y == 1) & (y_pred == 1))
    fp = np.sum((y == 0) & (y_pred == 1))
    fn = np.sum((y == 1) & (y_pred == 0))
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    return tp, fp, fn, recall

print("3. TP, FP, FN VA RECALL HISOBLARI:")
for t in [0.5, 0.3]:
    tp, fp, fn, rec = evaluate_metrics(p_vals, y_true, t)
    print(f"   Chegara {t} -> TP: {tp}, FP: {fp}, FN: {fn}, Recall: {rec:.2f}")
print()

# 4. BANK QAYSI CHEGARANI TANLASHI KERAK?
#  Javob: Bank 0.3 chegarasini tanlashi lozim.
#  Asoslash: Bank uchun eng qimmat xato bu kreditni qaytarmaydigan (1-sinf) mijozni aniqlay olmaslik (FN xatosi) hisoblanadi.
#  Chegara 0.5 da 1 ta firgarlik (FN=1) aniqlanmay qoladi.
#  Chegara 0.3 da FN=0 bo'ladi va barcha xavfli mijozlar to'liq aniqlanadi (Recall = 1.00).
#  Isboti sifatida, ishonchli mijozga rad javobini berish (FP) xarajati pulni butunlay yo'qotishdan ko'ra ancha kichikdir.

# 5. NEGA LOGISTIK REGRESSIYA ISHLATILADI?
# Chiziqli regressiya [-inf, +inf] oralig'ida cheksiz qiymatlar chiqaradi, bu binar tasniflashga (0 yoki 1) mos kelmaydi.
# Logistik regressiya esa Sigmoid yordamida natijani [0, 1] oralig'iga joylab, uni ehtimollik sifatida talqin qilish imkonini beradi.
```

```python
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

# 1. Yangi yo'lovchidan har bir nuqtagacha masofani hisoblang (2 xona aniqlikda).
X_train = np.array([
    [4, 4],
    [6, 7],
    [7, 5],
    [2, 3],
    [3, 2],
    [9, 9]
])
y_train = np.array(['E', 'K', 'K', 'E', 'E', 'K'])
yangi_yolovchi = np.array([5, 5])
masofalar = np.sqrt(np.sum((X_train - yangi_yolovchi) ** 2, axis=1))
print("1. YANGI YO'LOVCHIGACHA BO'LGAN MASOFALAR")
for i, d in enumerate(masofalar, start=1):
    print(f"Y{i} ({X_train[i-1][0]}, {X_train[i-1][1]}) [{y_train[i-1]}] gacha masofa: {d:.2f}")

# 2. k = 1, k = 3 va k = 5 bo'lganda bashorat qaysi tarif?
sorted_indices = np.argsort(masofalar)
print("2 ===")

for k in [1, 3, 5]:
    nearest_indices = sorted_indices[:k]
    nearest_labels = y_train[nearest_indices]

    values, counts = np.unique(nearest_labels, return_counts=True)
    predicted_class = values[np.argmax(counts)]

    print(f"k = {k} bo'lganda: Predict: {predicted_class}")
print()

# 3. "Safarlar soni" o'rniga "oylik xarajat (so'm)"
# Agar safarlar soni o'rniga juda katta qiymatga ega bo'lgan oylik xarajat (so'm)
# (masalan, 100 000) belgisi kiritilsa: KNN algoritmi faqat masofani o'lchagani uchun,
# masofa formulasida oylik xarajatning farqi ustunlikka ega bo'ladi. Kichik qiymatli
# masofa (km) o'z ta'sirini butunlay yo'qotadi. Tuzatish yo'li: Ma'lumotlarni masshtablash
# (Feature Scaling), masalan, StandardScaler (Z-score normalization) yoki MinMaxScaler
# yordamida har bir belgini bir xil holatga keltirish kerak.

# 4. Pipeline va GridSearch (cv=2 uchun k faqat 1 va 3 bo'lishi mumkin)
pipeline = Pipeline([('scaler', StandardScaler()), ('knn', KNeighborsClassifier())])
grid_search = GridSearchCV(pipeline, param_grid={'knn__n_neighbors': [1, 3]}, cv=2)
grid_search.fit(X_train, y_train)
print("4 pipeline")
print(f"Optimal k parametri (n_neighbors): {grid_search.best_params_['knn__n_neighbors']}")
print(f"Eng yaxshi o'quv aniqligi (score): {grid_search.best_score_:.2f}")
```

```python
import numpy as np

def gini(p0, p1):
    total = p0 + p1
    if total == 0:
        return 0
    return 1 - (p0/total)**2 - (p1/total)**2

# 1. Ildiz tugunning Gini qiymati:
g_root = gini(12, 8)
print(f"1. Ildiz tugunning Gini qiymati: {g_root:.4f}\n")

# 2. A bo'linish (Glukoza <= 125)
g_A_L = gini(11, 2)
g_A_R = gini(1, 6)
w_g_A = (13/20)*g_A_L + (7/20)*g_A_R
gain_A = g_root - w_g_A
print("2. A bo'linish natijalari:")
print(f"   - Chap bola Gini: {g_A_L:.4f}")
print(f"   - O'ng bola Gini: {g_A_R:.4f}")
print(f"   - Og'irlikli Gini: {w_g_A:.4f}")
print(f"   - Information Gain (Yutuq): {gain_A:.4f}\n")

# 3. B bo'linish (Yosh <= 45)
g_B_L = gini(7, 3)
g_B_R = gini(5, 5)
w_g_B = (10/20)*g_B_L + (10/20)*g_B_R
gain_B = g_root - w_g_B
print("3. B bo'linish natijalari:")
print(f"   - Og'irlikli Gini: {w_g_B:.4f}")
print(f"   - Information Gain (Yutuq): {gain_B:.4f}\n")

# 4. Qaror daraxtining tanlovi
print("4. Daraxt qaysi bo'linishni tanlaydi?")
print(f"   Javob: A bo'linishni tanlaydi. Chunki A ning Information Gain qiymati yuqoriroq ({gain_A:.4f} > {gain_B:.4f}).\n")

# 5. Overfitting muammosi va uning oldini olish
print("5. Muammo: Overfitting (O'ta moslashish).")
print("   Uni kamaytiradigan parametrlar: 1. max_depth, 2. min_samples_leaf\n")

# 6. Shifokor tushunadigan tilda oddiy qoida
print("6. A bo'linish o'ng bargining qoidasi:")
print("   Agar bemorning glukoza miqdori 125 dan yuqori bo'lsa (Glukoza > 125), unga diabet xavfi yuqori deb baholanadi.")
```

```python
import numpy as np
from sklearn.tree import DecisionTreeClassifier, export_text

data = [
    *([[100, 30], 0] for _ in range(7)),
    *([[100, 30], 1] for _ in range(1)),
    *([[100, 50], 0] for _ in range(4)),
    *([[100, 50], 1] for _ in range(1)),
    *([[150, 30], 1] for _ in range(2)),
    *([[150, 50], 0] for _ in range(1)),
    *([[150, 50], 1] for _ in range(4)),
]

X = np.array([row[0] for row in data])
y = np.array([row[1] for row in data])
#  1  gini hisoblash
def gini(counts):
    total = sum(counts)
    return 1.0 - sum((c / total) ** 2 for c in counts)

root_gini = gini([12, 8])

gini_a_left = gini([11, 2])
gini_a_right = gini([1, 6])
weighted_gini_a = (13 / 20) * gini_a_left + (7 / 20) * gini_a_right
gain_a = root_gini - weighted_gini_a

gini_b_left = gini([7, 3])
gini_b_right = gini([5, 5])
weighted_gini_b = (10 / 20) * gini_b_left + (10 / 20) * gini_b_right
gain_b = root_gini - weighted_gini_b

print(f"Ildiz Gini: {root_gini:.4f}")
print(f"A bo'linish: Chap={gini_a_left:.4f}, O'ng={gini_a_right:.4f}, Og'irlikli={weighted_gini_a:.4f}, Gain={gain_a:.4f}")
print(f"B bo'linish: Chap={gini_b_left:.4f}, O'ng={gini_b_right:.4f}, Og'irlikli={weighted_gini_b:.4f}, Gain={gain_b:.4f}")

dt = DecisionTreeClassifier(criterion='gini', max_depth=1)
dt.fit(X, y)

print("\n" + export_text(dt, feature_names=['Glukoza', 'Yosh']))
```
