```python
import numpy as np
def hosila(f, x, h=1e-5):
  return (f(x+h) - f(x)) / h
def kv(x): return x**2
numbers = np.linspace(-1, 1, 7)
for x in numbers:
  print(x, '->', hosila(kv, x))
```

```python
import sympy as sp

x = sp.symbols('x')

print(sp.diff(x**2))  # 2*x
print(sp.diff(x**3))  # 3*x**2
print(sp.diff(3*x**2 + 5*x - 7))  # 6*x + 5
# bu yerda diff berilgan functiondan hosila olib beryapti

# qiymatini topamiz: f'(2) = ?
fx = sp.diff(3*x**2 + 5*x - 7)
print(fx.subs(x, 2))  # 17
# yani hosila olgandan keyin, hisoblab berdi. subs orqali.
```

```python
import sympy as sp

x = sp.symbols('x')

print(sp.diff(x**4-2*x**2+9))
print(sp.diff(7*x**3+x-100))
print(sp.diff(((x**5)/5)-x))
print(sp.diff(1/x**2))

x = sp.symbols('x')
print(sp.solve(x**4-1, x))
```

```python
def gradient(f, x, y, h=1e-5):
  gx = (f(x+h, y) - f(x, y)) / h
  gy = (f(x, y+h) - f(x, y)) / h
  return np.array([gx, gy])

def f(x, y): return x**2 + y**2
print(gradient(f, 1, 2)) # [2, 4.]
print(gradient(f, 0, 0)) # [0, 0]

# (0, 0) da gradient NOL vektor -
# <<kosa>> tubidamiz
```

```python
def gradient(f, x, y, h=1e-5):
  gx = (f(x+h, y) - f(x, y)) / h
  gy = (f(x, y+h) - f(x, y)) / h
  return np.array([gx, gy])

def f(x, y): return ((x-3)**2 + (y+1)**2)
print(gradient(f, 0, 0))
print(gradient(f, 3, -1))
print(gradient(f, 5, 5))
```
