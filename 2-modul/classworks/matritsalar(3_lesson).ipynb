```python
I = np.eye(3)
print(I) # diagnali 1, qolgani 0

A = np.array([[5, 2, 1],
              [0, 3, 4],
              [1, 1, 2]])
print(np.allclose(A @ I, A)) # true
# A ni I ga ko'paytirdik -> A o'zi qoldi
```

```python
A = np.array([[3, 2],
              [4, 3]])
print(np.linalg.det(A))
A_inv = np.linalg.inv(A)
print(A_inv)
print(np.around(A @ A_inv))
```

```python
import numpy as np

def checkInv(A):
    a = np.linalg.det(A)
    if np.isclose(a, 0) or A.shape[0] != A.shape[1]: # Use np.isclose for robust comparison with zero
        print(f"Determinant nolga teng bo'lishi mumkin emas! Determinant: {a}")
        return None # Return None when inverse doesn't exist
    else:
        return np.linalg.inv(A)

A = np.array([[3, 2], [6, 4]])
A_inv_result_A = checkInv(A)
# No need to print A_inv_result_A here as the function prints the message

B = np.array([[1, 2], [3, 4]])
A_inv_result_B = checkInv(B)
print(A_inv_result_B)

# Only perform matrix multiplication if an inverse was returned
if A_inv_result_A is not None:
    print(np.allclose(A @ A_inv_result_A, np.eye(A.shape[0])))
else:
    print("Cannot check np.allclose for A, as its inverse does not exist.")
```
