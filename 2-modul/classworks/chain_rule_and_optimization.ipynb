```python
# y = (2 * x - 3)**2
# y(4) = 25 -> g(4) = u = 5 -> y(u) = 25
# y(0) = 9 -> g(0) = u = 3 -> y(u) = 9
# y(1) = 1 -> g(1) = u = (-1) -> y(u) = 1
# dy/dx = dy/du * du/dx     zanjir hosilasi
```

```python
def g(x): return 3*x + 1
def f(u): return u ** 2
def y(x): return f(g(x))  # murakkab funcsiya

h = 0.0001
x0 = 2
numeric_dydx = (y(x0 + h) - y(x0)) / h

chain_rule_dydx = 2 * g(x0) * 3 # dy/du *
print(f"numeric: {numeric_dydx:.3f}")
print(f"chain rule: {chain_rule_dydx}")



import sympy as sp
x = sp.symbols("x")
y = (3*x + 1)**2

dydx = sp.diff(y, x)
print(f"dy/dx = {dydx}")
print(f"dy/dx = {dydx.subs(x, x0)}")
```

```python
# numeric orqali
def g(x): return x**2 + 3*x
def f(u): return u ** 2
def y(x): return f(g(x))  # murakkab funcsiya

h = 0.0001
x0 = 1
numeric_dydx = (y(x0 + h) - y(x0)) / h
print(f"numeric: {numeric_dydx:.3f}")

# sympy orqali
import sympy as sp
x = sp.symbols("x")
y = (x**2 + 3*x)**2

dydx = sp.diff(y, x)
print(f"dy/dx = {dydx}")
print(f"sympy = {dydx.subs(x, x0)}")
```

```python
x, w1, b1, target = 2, 0.5, 1, 10

h = w1 * x + b1 # qatlam: h = 0.5 * 2 + 1
a = h ** 2   # aktivatsiya: a = 2 ^ 2 = 4
loss = (a - target)**2   # xato: (4-10)^2 = 36

# chain rule: dLoss/dw1 = dLoss/da * da/dh * dh/dw1
dLoss_da = 2 * (a - target)
da_dh = 2 * h
dh_dw1 = x
dLoss_dw1 =dLoss_da * da_dh * dh_dw1  # -96.0
print(f"dLoss_dw1 = {dLoss_dw1}")
```
