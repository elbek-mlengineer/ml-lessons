```python
def f(x):
  return np.sin(x) / x

# x ni 0 ga ikki tomondan yaqinlashtiramiz
for x in [-0.1, -0.01, -0.001, 0.001, 0.01, 0.1]:
  print(x, '->', f(x))

# bunda limit 1 ga teng
```

```python
import matplotlib.pyplot as plt

# def f(x): return np.abs(x)/x
def f(x): return 1/x

for x in [-0.01, -0.001, 0.001, 0.01]:
  print(x, '->', f(x))

# bunda limit 1 ga teng
xs = np.linspace(-2, 2, 400)
xs = xs[xs != 0]
plt.scatter(xs, f(xs), s=2)
plt.grid(True)
plt.xlim(-2, 2)
plt.ylim(-50, 50)
plt.show()
```

```python
def f(x):
  return (x+np.abs(x))/2

# x ni 0 ga ikki tomondan yaqinlashtiramiz
for x in [-0.1, -0.01, -0.001, 0.001, 0.01, 0.1]:
  print(x, '->', f(x))

xs = np.linspace(-2, 2, 400)
xs = xs[xs != 0]
plt.scatter(xs, f(xs), s=2)
plt.grid(True)
plt.show()
```

```python
fig, axes=plt.subplots(1, 3, figsize=(13, 4))

x = np.linspace(-3, 3, 600)
x= x[np.abs(x) > 1e-6]

x1=x[np.abs(x-2)> 1e-6]
axes[0].scatter(x1, (x1**2-4)/(x1-2), s=2)
axes[0].grid(True)
axes[0].scatter(2, 4, s=500, facecolors='none', edgecolors='r')

axes[1].scatter(x, np.abs(x)/x, s=2)
axes[1].grid(True)
axes[1].scatter(0, 1, s=500, facecolors='none', edgecolors='r')
axes[1].scatter(0, -1, s=500, facecolors='none', edgecolors='r')

axes[2].scatter(x, 1/x, s=2); axes[2].set_ylim(-20, 20)
axes[2].grid(True)
plt.show()
```

```python
def qiyalik(h):
  return ((3+h)**2 - 9) / h

for h in [0.1, 0.01, 0.001, 1e-6, -0.0001]:
  print(h,'->', qiyalik(h))

# limit = 6 <- bu Hosila bo'ladi!
```
