```python
#  (1/n)(wxi-yi)**2
import numpy as np
x = np.array([1, 2, 3])
y = np.array([3, 6, 9])
def cost(w):
  pred = w * x
  return np.mean((pred - y)**2)

for i in [1, 2, 3, 4]:
  print(f"w= {i}, cost -> {cost(i)}")
```

```python
import numpy as np
x = np.array([1, 2, 3])
y = np.array([3, 6, 9])
def gradient(w):
  pred = w * x
  return np.mean(2*x*(pred-y))

for i in [1, 2, 3, 4, 5]:
  print(f"w= {i}, gradient -> {gradient(i)}")

```

```python
import matplotlib.pyplot as plt

w=0
lr=0.05

w_history = [w]
cost_history = [cost(w)]

for step in range(6):
    g = gradient(w)
    w = w - lr * g
    w_history.append(w)
    cost_history.append(cost(w))
    print(f"{step+1}: w = {w:.4f}, cost={cost(w):.4f}")

w_range = np.linspace(-1, 7, 200)
cost_range = [cost(val) for val in w_range]

plt.figure(figsize=(5, 4))
plt.plot(w_range, cost_range, color="green")
plt.scatter(w_history, cost_history, color="red", zorder=5)
plt.grid(True)
plt.show()
```
