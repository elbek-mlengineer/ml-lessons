```python
from sklearn import datasets, svm
from sklearn.model_selection import train_test_split

#1. Ma'lumotlarni yuklash
X, y = datasets.load_iris(return_X_y=True)
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.2)
# 2. Modelni yaratish
model = svm.SVC(kernel='rbf', C=1.0, gamma='scale')
# 3. O'rnatish
model.fit(X_tr, y_tr)
# 4. Baholash (0 dan 1 gacha aniqlik)
print(model.score(X_te, y_te))
```

```python
import keras
from keras import layers
# 1. Ma'lumot: qo'lda yozilgan raqamlar (28x28)
(x_tr, y_tr), (x_te, y_te) = keras.datasets.mnist.load_data()
x_tr, x_te = x_tr / 255.0, x_te / 255.0
# 2. Model: qatlamlarni yig'amiz
model = keras.Sequential([
    layers.Input(shape=(28, 28)), layers.Flatten(),
    layers.Dense(128, activation="relu"),
    layers.Dense(64, activation="relu"),
    layers.Dense(10, activation="softmax")])
# 3-5. Kompilyatsiya, o'rgatish, baholash
model.compile(optimizer="adam", metrics=["accuracy"],
              loss="sparse_categorical_crossentropy")
# early stopping qilyapti. yani
early_stopping = keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=3,
    restore_best_weights=True,
    verbose=1,
)
model.fit(x_tr, y_tr, epochs=10, validation_split=0.1, callbacks=[early_stopping])
model.evaluate(x_te, y_te)
```
