```python
import numpy as np
import matplotlib.pyplot as plt
def f(x): return x
xato = []
def left_riemann(f, a, b, n):
  dx = (b-a)/n
  xs = np.linspace(a, b, n, endpoint=False)
  return np.sum(f(xs)) * dx

for n in [4, 8, 20]:
  xato.append(abs(8 - left_riemann(f, 0, 4, n)))
  print(f"n={n}: {left_riemann(f, 0, 4, n):.3f} xato: {8 - left_riemann(f, 0, 4, n)}")
plt.plot(np.linspace(0, 4, 20), f(np.linspace(0, 4, 20)) , label="f(x)")
plt.plot([4, 8, 20], xato, label="xatolik")
plt.legend()

plt.grid(True)
plt.show()
```

```python
# trapetsiya yuzasini topish uchun
xs = np.linspace(0, 3, 11)
ys = f(xs)
natija = np.trapezoid(ys, xs)
print(f"trapetsiya (n=10): {natija:.4f}")
```

```python
# hosiladan boshlang'ich olish
import sympy as sp
x = sp.symbols('x')
F1 = sp.integrate(x**3, x)
print(f"x^3dx = {F1} + C")

F2 = sp.integrate(3*x**2, x)
print(f"3x^2dx = {F2} + C")

F3 = sp.integrate(x**4, x)
print(f"x^4dx = {F3} + C")

F4 = sp.integrate(7, x)
print(f"7dx = {F4} + C")

F5 = sp.integrate(x**5, x)
print(f"x^5dx = {F5} + C")

F6 = sp.integrate(4*x**3, x)
print(f"4*x^3dx = {F6} + C")
```

```python
# 1-usul: qo'lda hisoblash(newton-leibniz)
F = lambda x: x**3/3
qolda = F(3)-F(0)
print(f"Qo'lda: {qolda}")
# 2-usul: sympy bilan avtomatik
sympy_natija = sp.integrate(x**2, (x, 0, 3))
print(f"Sympy natija: {sympy_natija}")
```

```python
def f(x):
  return 2*x+3
F = lambda x: x**2 + 3*x
qolda = F(4)-F(0)
print(f"Qo'lda: {qolda}")

sympy_natija = sp.integrate(f(x))
print(f"antihosila: {sympy_natija} + C")

sympy_natija = sp.integrate(f(x), (x, 0, 4))
print(f"yuza: {sympy_natija}")

xs = np.linspace(0, 4, 9)
ys = f(xs)
natija = np.trapezoid(ys, xs)
print(f"trapetsiya (n=10): {natija:.4f}")
```
