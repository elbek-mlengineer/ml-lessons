```python
import seaborn as sns
import matplotlib.pyplot as plt
data1 = sns.load_dataset('tips')
sns.histplot(data=data1, x='total_bill')
plt.show()
```

```python
tips = sns.load_dataset('tips')
for context in ["paper", "notebook", "talk", "poster"]:
  sns.set_theme(context = context)
  sns.histplot(data=tips, x='total_bill')
  plt.title(f"Context: {context}")
  plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
tips = sns.load_dataset('tips')

sns.histplot(
    data=tips,
    x="total_bill",
    bins=30,
    kde=True
)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
tips = sns.load_dataset('tips')
sns.kdeplot(
    data=tips,
    x="total_bill",
    hue="sex",
    fill=True,
    alpha=0.5
)
plt.title("kdeplot example")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')
sns.boxplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex"
)
plt.title("boxplot example")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
tips = sns.load_dataset('tips')
sns.violinplot(
    data=tips,
    x="total_bill",
    y="day",
    hue="sex",
    cut=1,
    scale="width"
)
plt.title("violin example")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')
sns.stripplot(
    data= tips,
    x="day",
    y="total_bill",
    jitter=True,
    alpha=0.5
)
plt.title("stripplot example")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')
sns.swarmplot(
    data=tips,
    x="day",
    y="total_bill"
)
plt.title("swarmplot graph")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')
sns.regplot(
    data=tips,
    x="total_bill",
    y="tip",
    ci=95
)
plt.title("regplot example")
plt.show()
#
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')
sns.jointplot(
    data=tips,
    x="total_bill",
    y="tip",
    kind="reg"
)
plt.show()
```
