```python
import kagglehub
import matplotlib.pyplot as plt
import pandas as pd

path = kagglehub.dataset_download("iamsouravbanerjee/customer-shopping-trends-dataset")

df = pd.read_csv(path + "/shopping_trends.csv")
age = df.groupby('Age')['Purchase Amount (USD)'].mean().reset_index()

age = age.sort_values('Age')

plt.figure(figsize=(15, 6))
plt.plot(age['Age'], age['Purchase Amount (USD)'],
         color="red",
         linestyle="--",
         linewidth=2,
         marker="o"
)
plt.title("Yoshga qarab o'rtacha xarid summasi")
plt.xlabel("Yosh")
plt.ylabel("O'rtacha xarid summasi (USD)")
plt.grid(True)
```

```python
import kagglehub
import matplotlib.pyplot as plt
import pandas as pd

path = kagglehub.dataset_download("iamsouravbanerjee/customer-shopping-trends-dataset")

df = pd.read_csv(path + "/shopping_trends.csv")
category_counts = df['Category'].value_counts().reset_index()
category_counts.columns = ['Category', 'Count']
category_counts = category_counts.sort_values('Count', ascending=True)

plt.figure(figsize=(12, 8))
plt.barh(category_counts['Category'], category_counts['Count'],
         color="skyblue",
         edgecolor="darkblue",
         linewidth=1.5
)
plt.title("Mahsulot kategoriyalari bo'yicha sotib olingan mahsulotlar soni")
plt.xlabel("Sotib olingan mahsulotlar soni")
plt.ylabel("Kategoriya")
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.show()
```

```python
import kagglehub
import matplotlib.pyplot as plt
import pandas as pd

path = kagglehub.dataset_download("iamsouravbanerjee/customer-shopping-trends-dataset")

df = pd.read_csv(path + "/shopping_trends.csv")
data = df.sample(n=500, random_state=42)

plt.figure(figsize=(12, 8))
plt.scatter(data['Age'], data['Purchase Amount (USD)'],
            color="skyblue",
            alpha=0.8,
            s=50,
            edgecolor='darkblue',
            linewidth=1.5
)
plt.title("Yosh va xarid summasi o'rtasidagi munosabat (Namuna asosida)")
plt.xlabel("Yosh")
plt.ylabel("Xarid summasi (USD)")
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import kagglehub

path = kagglehub.dataset_download("iamsouravbanerjee/customer-shopping-trends-dataset")
df = pd.read_csv(path + "/shopping_trends.csv")

age = df.groupby('Age')['Purchase Amount (USD)'].mean().reset_index()
age = age.sort_values('Age')

plt.figure(figsize=(15, 6))
sns.lineplot(data=age, x='Age', y='Purchase Amount (USD)',
             color="red",
             linestyle="--",
             linewidth=2,
             marker="o"
)
plt.title("Yoshga qarab o'rtacha xarid summasi (Seaborn)", fontsize=14, fontweight="bold")
plt.xlabel("Yosh", fontsize=12)
plt.ylabel("O'rtacha xarid summasi (USD)", fontsize=12)
plt.grid(True, linestyle="--", alpha=0.7)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import kagglehub

path = kagglehub.dataset_download("iamsouravbanerjee/customer-shopping-trends-dataset")
df = pd.read_csv(path + "/shopping_trends.csv")

category_counts = df['Category'].value_counts().reset_index()
category_counts.columns = ['Category', 'Count']
category_counts = category_counts.sort_values('Count', ascending=True)

plt.figure(figsize=(12, 8))
sns.barplot(x='Count', y='Category', data=category_counts,
            color="skyblue",
            edgecolor="darkblue",
            linewidth=1.5
)
plt.title("Mahsulot kategoriyalari bo'yicha sotib olingan mahsulotlar soni (Seaborn)", fontsize=14, fontweight="bold")
plt.xlabel("Sotib olingan mahsulotlar soni", fontsize=12)
plt.ylabel("Kategoriya", fontsize=12)
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.show()
```
