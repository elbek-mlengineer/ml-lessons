```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom
```

```python
n = 20
p = 0.3
mu = n * p # matematik kutish

var = n * p * (1 - p)

sigma = np.sqrt(var)

print("E[X] =", mu)
print("Var(X) =", var)
print("σ =", sigma)
```

```python
lower = mu - sigma
upper = mu + sigma

print("μ - σ =", lower)
print("μ + σ =", upper)
print("μ ± σ oralig'i:", lower, upper)
```

```python
x_vals = np.arange(0, n + 1)
probs = binom.pmf(x_vals, n, p)

plt.figure(figsize=(10, 5))

plt.bar(
    x_vals,
    probs,
    color="#2ECB9B",
    alpha=0.85,
    edgecolor="white",
    width=0.7
)

plt.xlabel("X qiymati")
plt.ylabel("P(X = x)")
plt.title("Binomial PMF — B(20, 0.3)")

plt.xticks(x_vals)
plt.ylim(0, 0.22)

plt.tight_layout()
plt.show()
```

```python
from scipy import stats
import matplotlib.pyplot as plt
import numpy as np

mu1, sigma1 = 100, 10
dist1 = stats.norm(mu1, sigma1)

mu2, sigma2 = 100, 25
dist2 = stats.norm(mu2, sigma2)

x = np.linspace(20, 180, 300)

plt.plot(x, dist1.pdf(x), color="#2ECB9B",
         lw=2.5, label="N(100, 10²)")

plt.plot(x, dist2.pdf(x), color="blue",
         lw=2.5, label="N(100, 25²)")

plt.axvline(mu1, color="red", linestyle="--")

# 1-taqsimot uchun
plt.axvline(mu1 + sigma1, color="orange",
            linestyle=":")

plt.axvline(mu1 - sigma1, color="orange",
            linestyle=":")

# 2-taqsimot uchun
plt.axvline(mu2 + sigma2, color="purple",
            linestyle=":")

plt.axvline(mu2 - sigma2, color="purple",
            linestyle=":")

plt.fill_between(
    x,
    dist1.pdf(x),
    where=(x >= mu1-sigma1) & (x <= mu1+sigma1),
    alpha=0.3,
    color="#2ECB9B"
)

plt.fill_between(
    x,
    dist2.pdf(x),
    where=(x >= mu2-sigma2) & (x <= mu2+sigma2),
    alpha=0.15,
    color="blue"
)

plt.title("Ikkita Normal Taqsimotni Taqqoslash")
plt.xlabel("X")
plt.ylabel("Zichlik")

plt.legend()
plt.show()
```

```python
from scipy import stats
import numpy as np

n, p = 10, 0.4

np.random.seed(42)
X = stats.binom.rvs(n, p, size=100000)

Y = 3 * X + 7

var_Y_1 = np.var(Y)

var_X = np.var(X)
var_Y_2 = 3**2 * var_X

print("X = Binomial(10, 0.4)")
print("Y = 3X + 7")

print(f"Var(X) = {var_X:.4f}")
print(f"Var(Y) — 1-usul = {var_Y_1:.4f}")
print(f"Var(Y) — 2-usul = {var_Y_2:.4f}")
# ikkala usulda ham bir xil natija chiqyapti
```
