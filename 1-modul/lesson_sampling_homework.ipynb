```python
import pandas as pd
import random

# Random natija har safar bir xil chiqishi uchun
random.seed(42)

names = [
    "Ali", "Vali", "Hasan", "Husan", "Sardor",
    "Jasur", "Akmal", "Bekzod", "Aziz", "Bobur"
]

data = []

# 3 ta kurs, har birida 100 tadan student
for kurs in [1, 2, 3]:
    for i in range(1, 101):
        data.append({
            "student_id": (kurs - 1) * 100 + i,
            "ism": random.choice(names),
            "kurs": kurs,
            "yosh": random.randint(18, 24)
        })
df = pd.DataFrame(data)

# print(df)
users = df.groupby('kurs').apply(lambda x: x.sample(10))
print(users)
```

```python
import numpy as np

population = np.arange(1, 101)

N = len(population)
n = 10

k = N // n

sample = population[::k]

print("k =", k)
print("Sample =", sample)
```

```python
import numpy as np

sample = [60, 70, 80, 90, 100]

sd = np.std(sample, ddof=1)
n = len(sample)

se = sd / np.sqrt(n)

print("Standard Deviation:", sd)
print("Sample size:", n)
print("Standard Error:", se)
```
